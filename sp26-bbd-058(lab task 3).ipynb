```python
# question 1
# simple counter
i = 1
while i <= 10:
    print(i)
    i += 1
```

```python
# question 2
# countdown
n = int(input("Enter starting number: "))
while n >= 0:
    print(n)
    n -= 1
```

```python
# question 3
# sum of first N numbers
N= int(input("Enter a positive integer N: "))
total = 0
i = 1
while i <= N:
    total += i
    i += 1
print(f"Sum = {total}")
```

```python
# question 4
# multiplication table
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1
```

```python
# question 5
# Input Validation
num = float(input("Enter a positive number: "))
while num <= 0:
    print("Invalid! Try again.")
    num = float(input("Enter a positive number: "))
print(f"You entered: {num}")
```

```python
# question 6
# Password Check
password = "admin123"
attempt = input("Enter password: ")
while attempt != password:
    print("Wrong password. Try again.")
    attempt = input("Enter password: ")
print("Access granted!")
```

```python
# question 7
# Count Digits
num = int(input("Enter a number: "))
count = 0
temp = abs(num) if num != 0 else 1
while temp > 0:
    temp //= 10
    count += 1
print(f"Digit count: {count}")
```

```python
#question 8
#Running Total Until 0
total = 0
num = float(input("Enter a number (0 to stop): "))
while num != 0:
    total += num
    num = float(input("Enter a number (0 to stop): "))
print(f"Total: {total}")
```

```python
# question 9
# First Multiple of 7 Over 100
n = 101
while n % 7 != 0:
    n += 1
print(f"First multiple of 7 over 100: {n}")
```

```python
# question 10
# Number Guessing Game
secret = 25
attempts = 0
guess = int(input("Guess the number: "))
while guess != secret:
    attempts += 1
    if guess > secret:
        print("Too high!")
    else:
        print("Too low!")
    guess = int(input("Guess again: "))
attempts += 1
print(f"Correct! You got it in {attempts} attempts.")
```
